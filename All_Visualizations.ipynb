```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

I. Distribution plots

1.Histogram

```python
np.random.seed(3)
marks = np.random.normal(65, 12, 300)

plt.figure(figsize=(100,50),facecolor="blue")

ax= plt.axes()
ax.set_facecolor("yellow")

plt.hist(marks, bins =15, edgecolor="purple",color="green",linewidth=10,linestyle=":",alpha=0.5)
plt.grid(True)
plt.title("Marks Distribution",color="red",fontsize=300)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.xticks(fontsize=30,rotation=180)





plt.show()
```

```python
sns.histplot(marks,bins=15)
plt.grid(color="gray",linestyle="-.",alpha=0.5,linewidth=1)
plt.show()
```

02. box plot

```python
np.random.seed(5)
subjects = np.repeat(["Math", "Science", "English", "ICT"], 25)
scores   = np.concatenate([
    np.random.normal(72, 8, 25),
    np.random.normal(65, 10, 25),
    np.random.normal(78, 6, 25),
    np.random.normal(70, 12, 25)])

df = pd.DataFrame({"Subject": subjects, "Score": scores})

df.boxplot(column="Score", by="Subject")
plt.show()

sns.boxplot(x=subjects, y=scores)
plt.xticks(rotation=45,size=30, color="Blue")
plt.show()

sns.boxplot(x="Subject", y="Score", data = df)
```

3. Violin Plot


```python
sns.violinplot(x=subjects,y=scores)
```

```python
x = np.linspace(1,5,100)

fig, ax = plt.subplots(2, 2, figsize=(10, 8))
fig.suptitle('Subplots with Grid', fontsize=16)

# Apply grid to each subplot and plot the functions
ax[0, 0].plot(x, np.sin(x), 'r-.', label='sin(x)')
ax[0, 0].set_title('Sine Function')
ax[0, 0].legend()
ax[0, 0].grid(True)

ax[0, 1].plot(x, np.cos(x), 'g--', label='cos(x)')
ax[0, 1].set_title('Cosine Function')
ax[0, 1].legend()
ax[0, 1].grid(True)

ax[1, 0].plot(x, np.tan(x), 'y-', label='tan(x)')
ax[1, 0].set_title('Tangent Function')
ax[1, 0].legend()
ax[1, 0].grid(True)

ax[1, 1].plot(x, np.sinc(x), 'c.-', label='sinc(x)')
ax[1, 1].set_title('Sinc Function')
ax[1, 1].legend()
ax[1, 1].grid(True)

plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))

ax1.plot(months, sales, marker="D", markersize=8, markerfacecolor="red", markeredgecolor="black")
ax1.set_title('Months vs. Sales')
ax1.set_xlabel('Month')
ax1.set_ylabel('Sales')
ax1.grid(True)

ax2.plot(sales, months, marker="D", markersize=8, markerfacecolor="red", markeredgecolor="black")
ax2.set_title('Sales vs. Months')
ax2.set_xlabel('Sales')
ax2.set_ylabel('Month')
ax2.grid(True)

plt.tight_layout()
plt.show()
```
